# fish-oot sweep worker (Colab)

Runs part of a sweep in this session and pushes the results. See `fishmod/sweep/README.md`.
Run the cells in order. Nothing here stores your token or your ROM outside this session.

```python
# 1. A GitHub token that can read and write the repository (a fine-grained token with Contents: read and write).
import getpass, os, subprocess
token = getpass.getpass("GitHub token: ")
repo = "RastrumQuill/fish-oot"
if not os.path.isdir("/content/fish-oot"):
    subprocess.run(["git", "clone", "-q", f"https://x-access-token:{token}@github.com/{repo}", "/content/fish-oot"], check=True)
subprocess.run(["git", "-C", "/content/fish-oot", "pull", "-q"], check=True)
print("repository ready")
```

```python
# 2. Your ROM (Ocarina of Time NTSC 1.2 (US), any byte order). Either upload it here...
from google.colab import files
up = files.upload()
ROM = "/content/" + next(iter(up))
# ...or comment the three lines above out and read it from your Drive instead:
# from google.colab import drive; drive.mount("/content/drive"); ROM = "/content/drive/MyDrive/oot.z64"
print(ROM)
```

```python
# 3. The worker. The first run installs packages and builds the emulator (about 20 minutes here).
#    It stops when the queue is empty; stopping the cell stops it (its unfinished units are freed after 8 hours).
!cd /content/fish-oot && fishmod/sweep/worker.sh --rom "$ROM" --name colab --parallel 2
```
